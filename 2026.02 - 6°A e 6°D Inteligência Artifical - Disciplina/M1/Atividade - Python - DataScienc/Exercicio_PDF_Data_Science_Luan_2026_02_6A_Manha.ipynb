#Exercicio do NumPy - 7. Crie dois arrays de números aleatórios e encontre a correlação entre eles.

---

#Justificativa
Importei a biblioteca do NumPy, criando 2 array aleatórios utilizando a função random randint() para que os números aleatórios sejam inteiros e criando as variáveis "media1" e "media2" para que dentro desses 2 arrays tirasse a média deles usando a função do NumPy mean(), e subtraio a media pelo array para obter o desvio deles, e então eu somo o desvio obetido anteiormente e com isso "n1 = len(array1) -1" o denominador - (n-1) * desvio-padrao - ** o n-1 desfaz a soma e volta para o desvio padrao, e então utilizo o np.std com ddof=1 garante o uso de n-1 no cálculo. E para o coeficiente de Correlação de Pearson usei a formula r = Covariância / (desvio_padrao1 * desvio_padrao2) isso para obter a correlação entre os array.

```python
import numpy as np


#Criando os arrays com numeros aleatorios
array1 = np.random.randint(1,10,size=10)
array2 = np.random.randint(20,30,size=10)

#pegando a media dos arrays - provavel localizacao dos elementos
media1 = np.mean(array1)
media2 = np.mean(array2)

#subtrair a media de cada elemento para ter o desvio
desvio1 = array1 - media1
desvio2 = array2 - media2

#soma dos elementos dos desvio
numerador = np.sum(desvio1 * desvio2)


#denominador - (n-1) * desvio-padrao - ** o n-1 desfaz a soma e volta para o desvio padrao
# O np.std com ddof=1 garante o uso de n-1 no cálculo
n1 = len(array1) -1

solucao1 = np.std(array1, ddof=1)
solucao2 = np.std(array2, ddof=1)


#coeficiente de Correlação de Pearson - formula
# r = Covariância / (desvio_padrao1 * desvio_padrao2)

r = numerador / (n1*solucao1*solucao2)


#formula : correlacao de pearson = quando x e y variam junos * quanto cada um varia sozinho




print(array1)
print(array2)
print("correlacao dos coeficientes: {}".format(r))
```

#Exercicio do NumPy - 19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

---

#Justificativa

Importei a biblioteca NumPy para armazenar numeros a lista, comecei gerando dois conjuntos de 20 números inteiros aleatórios entre 1 e 100 usando o comando randint(). Para achar o produto escalar, multipliquei os dois arrays elemento por elemento e usei a função sum() para somar tudo, seguindo o conceito de multiplicação cruzada de arrays. Depois, calculei o comprimento de cada array, elevando os valores ao quadrado, somando com sum() e tirando a raiz quadrada com a função sqrt. Depois dividi o "produtoEscala"r pelo resultado da multiplicacao das duas normas para encontrar o cosseno do angulo entre os arrays e por fim, usei o comando arccos() para descobrir o tamanho desse ângulo em radianos e a função degrees() para transformar essa medida em graus.

```python
import numpy as np

array1 = np.random.randint(1,100, size=20)

array2 = np.random.randint(1,100, size=20)

print(array1)
print(array2)
produtoEscalar = np.sum(array1 * array2)



# norma (magnitude/"comprimento") de cada vetor
#tamanho do vetor no espaco
norma1 = np.sqrt(np.sum(array1 ** 2))
norma2 = np.sqrt(np.sum(array2 ** 2))

# cosseno do ângulo = produto escalar normalizado
cosTheta = produtoEscalar / (norma1 * norma2)


# converte de cosseno para ângulo
anguloRadiano = np.arccos(cosTheta)
anguloGraus = np.degrees(anguloRadiano)


print("Angulo Radiano: {}".format(anguloRadiano))
print("Angulo Graus: {}".format(anguloGraus))

```

#Justificativa

A função recebe um único parâmetro valor e desempacota x e y dessa variável, porque o minimize só aceita um vetor com as variáveis agrupadas, não valores separados. A variável partida define o ponto inicial de onde o algoritmo começa a busca, nesse caso "[0,0]" O resultado é armazenado em resultado, e ao final são impressos três dados: se o algoritmo convergiu com sucesso, as coordenadas x e y do ponto mínimo encontrado, e o valor mínimo da função nesse ponto.

```python
import numpy as np
from scipy.optimize import minimize

def funcao(valor):
  x,y = valor # pq o minimize aceita apenas uma matriz ou vetor e não uma saida de 2 valores como estava
  return (x-2)**2+(y-3)**2

#definimos o palpite inicial
partida = [0,0]

# O algoritmo calcula numericamente o caminho até o ponto mais baixo
resultado = minimize(funcao,partida)

print(resultado.success)
print("minimo encontrado de x e y: {}".format(resultado.x))
print("valor minimo da função: {}".format(resultado.fun))
```

Defino erro como 1, e calculo lmenos (y menos o erro) e lmais (y mais o erro), que formam o intervalo de confianca. Uso plt.fill_between(x, lmenos, lmais) pra sombrear a area entre lmenos e lmais, mostrando visualmente a faixa de confianca em volta da linha. Uso plt.plot(x,y, color='black', label='y = sin(x)') pra plotar a linha principal do seno. Adiciono titulo, nomes dos eixos e grade com plt.grid(True).
No final uso plt.show() pra exibir o grafico.

```python
import numpy as np
import matplotlib.pyplot as plt

#Gerando dados
x = np.linspace(0,10,100)
y = np.sin(x)


#intervalo de confianca
erro = 1
lmenos = y - erro
lmais = y + erro

#adicionando sombreamento
plt.fill_between(x, lmenos, lmais)

#plota a linha principal
plt.plot(x,y, color='black', label='y = sin(x)')
plt.title("Grafico com intervalos de Confianca")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.grid(True)
plt.show()

```

Calculo y1 como seno de x, y2 como cosseno de x, e y3 como a soma dos dois. Uso plt.subplot(3,1,1) pra criar uma grade de 3 linhas e 1 coluna, e plotar o primeiro grafico (seno) nessa posicao. Repito o processo com plt.subplot(3,1,2) pra plotar o cosseno na segunda posicao, e plt.subplot(3,1,3) pra plotar a soma dos dois na terceira posicao. Cada um recebe seu titulo. Uso plt.tight_layout() pra ajustar o espacamento entre os graficos e evitar que titulos e eixos se sobreponham.
No final uso plt.show() pra exibir os tres graficos juntos.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) + np.cos(x)

plt.subplot(3,1,1) # 2 linhas,1 coluna, 1 grafico
plt.plot(x,y1)
plt.title("Grafico de seno")

plt.subplot(3,1,2) #2 linhas, 1 coluna, 2 grafico
plt.plot(x,y2)
plt.title("Grafico de cosseno")

plt.subplot(3,1,3) # 2 linhas, 2 coluna, 3 grafico
plt.plot(x,y3)
plt.title("Grafico de seno mais o cosseno")

plt.tight_layout() #ajustar o layout para evitar sobreposicao
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# gera 100 valores aleatorios pra x, y e z
x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

fig = plt.figure(figsize=(8,6))# cria a figura
ax1 = fig.add_subplot(projection='3d')# cria eixo com projecao 3d

ax1.scatter(x,y,z,c=z)#aqui ele plota os pontos em 3d, colorindo pelo valor de z
ax1.set_xlabel('Eixo x')
ax1.set_ylabel('Eixo y')
ax1.set_zlabel('Eixo z')
plt.show()
```

#Justificativa
Importo o numpy pra trabalhar com arrays, e o matplotlib.pyplot pra criar graficos. Crio a lista categorias com os nomes, e valores com numeros positivos e negativos. Uso "np.where(np.array(valores) >= 0, 'black', 'red')" pra criar a lista cores, definindo preto pros valores positivos e vermelho pros negativos. Uso "plt.bar(categorias,valores,color=cores)" pra criar o grafico de barras, cada uma com a cor correspondente. Uso "plt.axhline(0,color='black', linewidth=0.8,linestyle='--')" pra desenhar uma linha horizontal tracejada no valor zero, marcando a divisao entre positivo e negativo. ```python
import numpy as np
import matplotlib.pyplot as plt

categorias = ['Ado','Huntrix','Disney','Anime','Hexatombe'] # lista com as categorias

valores = [55,-30,44,-50,-12] # lista com valores positivos e negativos

cores = np.where(np.array(valores) >= 0, 'black', 'red') # define preto pros positivos e vermelho pros negativos

plt.bar(categorias,valores,color=cores)# cria grafico de barras com as cores definidas
plt.axhline(0,color='black', linewidth=0.8,linestyle='--')# linha horizontal no zero
plt.title("Grafico de Produtos legais ")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```

#Exercicio do Matplotlib - 20. Adicione uma grade com espaçamento personalizado a um gráfico de linha para y = x².

---
#Justificativa

Importo o numpy pra trabalhar com numeros e arrays, e o matplotlib.pyplot pra criar graficos. Uso "np.linspace(-5,10,100)" pra gerar 100 pontos entre -5 e 10, guardados em x, e calculo y como x ao quadrado. Uso "plt.plot(x,y,color="black", label="y = x²")" pra plotar a linha do grafico na cor preta, ja com label pra aparecer na legenda. Uso "plt.xticks(np.arange(-5,11,2))" e "plt.yticks(np.arange(0,101,10))" pra personalizar o espacamento das marcacoes nos eixos x e y, de 2 em 2 e de 10 em 10 respectivamente. Uso "plt.grid(color='gray', linestyle='--', linewidth=0.5)" pra adicionar a grade seguindo esse espacamento personalizado, na cor cinza e com linhas tracejadas. ```python
import numpy as np
import matplotlib.pyplot as plt


x = np.linspace(-5,10,100) # gera 100 pontos entre -5 e 10
y = x**2 # calcula y como x ao quadrado


plt.plot(x,y,color="black", label="y = x²") # plota a linha em preto com label
plt.xticks(np.arange(-5,11,2)) # define marcacoes do eixo x de 2 em 2
plt.yticks(np.arange(0,101,10)) # define marcacoes do eixo y de 10 em 10
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()
plt.title("Grafico de Linha para y=x² ")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```
