  - "if letra in cont:" - Aqui eu utilizei o operador in dentro do if para comparar que vê se deve incrementar um contador caso na letra existir.
  - "cont[letra] += 1 - cont[letra] = 1" - para contar cada letra se ela já for uma chave no vetor ela utiliza a atribuição composta "+=" e soma o seu valor atual + 1 e caso contrário no else ele cria uma nova chave e inicializa com o valor 1.
- "print(cont)" - e imprimo o contador para mostrar para o usuário.




```python

palavra =  "palavra" # criando a variável da palavra de caracteres

cont = {} # criando um vetor para receber chave e valor

for letra in palavra: # passar por cada letra da palavra / variável
  if letra in cont: # contando cada letra
    cont[letra] += 1 # as letras que já foram encontradas somam mais um a cada repetição
  else:
    cont[letra] = 1 # contando apenas 1 para a primeira vez que aparecem

print(cont)

```

- "if palavra in cont:" - Aqui eu utilizei o operador in dentro do if para comparar que vê se deve incrementar um contador caso na palavra existir.

 "print(cont)" - e imprimo o contador para mostrar para o usuário.

Pedi para que o usuário digitasse um número que ele escolhesse.
-  "primo = True"-  e decidi começar uma variável "primo" como true, para que ele continuasse verdadeiro durante o processamento
- "if numero < 2: e primo = False" - pensei que aqui em fazer um loop que todo o número que fosse menor que 2 ele fosse falso porque ele não seria primo
- "else: primo = True e for i in range(2, numero):"  aqui no else ele começaria como primo e separaria os primos, pelo for que ele itera sobre a sequência, que ele gera um iterador que começa em 2 até o valor de "numero" sem incluir ele mesmo e isso vai testar apenas os divisores de 2 e o número anterior digitado
- E no " if numero % i == 0:" - aqui eu pensei que de dentro desse for o resto da divisão do número por qualquer i for zero, isso iria provar que o número que possui um divisor além de 1 e dele mesmo, invalida ele como primo
 e utilizei o break para encerrar a repetição.


```python

numero = int(input("Digite um número: "))
primo = True # começando a variável com true

if numero < 2: # aqui todo numero menor que 2 não é primo
  primo = False
else:
  primo = True
  for i in range(2, numero): # cria um loop a partir do 2 até o número anterior do digitado pelo usuário.
    if numero % i == 0: # Caso o resto da divisão do número pelo i seja 0, o número não é considerado primo
      primo = False
      break


if primo:
  print("O número é primo")
else:
  print("O número não é primo")


```

##Exercicio 19: Implemente um jogo onde o usuário deve adivinhar um número gerado aleatoriamente pelo computador.

---

# Justificativa
Eu utilizei e importei a biblioteca "random" para que o número fosse gerado aleatoriamente pelo computador
 - Crio uma variável para que o usuário possa tentar adivinhar o número gerado pelo computador
- "numero = random.randint(0,50)" - aqui eu crio uma variável para receber o número aleatório do computador que ele recebe a função "random.randint(0,50)" que ele vai sortear um número inteiro aleatório dentro do intervalo especificado.
  - "while numero != escolhido:" - no laço while eu comparo se o número gerado pela função é diferente do que o usuário digitou se for verdadeiro o loop continua até ele acertar
  - "if numero == escolhido:" - e se o número escolhido pelo usuário for igual ao que foi gerado então é verdadeiro e eu imprimo uma mensagem de " Parabéns você acertou".
  - "elif numero > escolhido: " - e aqui eu pensei que se ele apenas ficar adivinhando não seria ideal então coloquei outra condição que se o número que ele digitasse estivesse perto ou longe do número gerado.
E deixo que ele tente de novo a cada erro que possa acontecer.





```python
import random # biblioteca para gerar um número aleatório

escolhido = int(input("Escolha um numero de 0 a 50: ")) # peço que o usuário escolha um numero

numero = random.randint(0,50) # faço com que a variável gere um número aleatório de 0 a 50
while numero != escolhido: # enquanto o numero aleatório for diferente do numero escolhido ele fica no loop
    if numero == escolhido: # se o número escolhido for igual ao numero aleatório ele acerta
      print("Parabens você acertou")
    elif numero > escolhido: # se for diferente ele fala se é maior ou menor
      print("o numero escolhido é maior")

    else:
      print("o numero escolhido é menor")

    escolhido = int(input("Tente de novo! Escolha um novo número entre 0 a 50: ")) # e toda vez que ele erra, pede para tentar de novo até acertar.

print("Parabens você acertou")

```

##Exercicio 22: Crie uma função para criptografar uma mensagem usando a cifra de César.

- "for letra in texto: e if letra.isalpha():" - Aqui o laço percorre a sequência nos caracteres da "letra" e o método ".isalpha()" verifica se o caractere pertence ao alfabeto que assim permite que na lógica de criptografia ignore símbolos e espaços.
- "ord(letra), chr(...) e números 65/97" - A função "ord()" converte o caractere em código numérico correspondente, enquanto o "chr()" ele faz o inverso, e aí que os valores 65 da base "A" e 97 da base "a" é utilizado para normalizar a posição da letra na tabela antes que a rotação aconteça.
- E o % 26 utilizei o 26 porque é o limite das letras do alfabeto criando um efeito de ciclo onde Z volta pra A.
- "texto_criptografado += letra_criptografada " - Fiz uma atribuição composta para que durante a mudança das strings ele criasse uma nova string combinando com a anterior.


```python

def crip(texto,chave):



  texto_criptografado = "" # iniciando a variável

  for letra in texto:
    if letra.isalpha(): # verificar se na letra contém apenas letras do alfabeto
      if letra.isupper(): # verificar se estão em maiusculas
        letra_criptografada = chr((ord(letra) - 65 + chave) % 26 + 65) # verifica a letra na tabela de codigo ASCII do computador e altera o valor da letra, recomeçando com 0, mantém o resultado dentro do limite de 26 letras do alfabeto + o seu valor
      else:
        letra_criptografada = chr((ord(letra) - 97 + chave) % 26 + 97) # tranforma a letra no numero correspondente no alfabeto, e com a chave move a letra para frente ou para tras para fazer a troca.
      texto_criptografado += letra_criptografada # junta a nova letra no final do texto final
    else:
      texto_criptografado += letra # ele mantém os caracteres que não são letras.
  return texto_criptografado

texto = input("Digite uma frase para criptografar: ")
chave = int(input("Digite a chave de criptografia: "))
result = crip(texto,chave)
print("Mensagem criptografada: {}".format(result))





```

No segundo, verifico se todos os números são iguais entre si, como "11111111111". Caso qualquer uma dessas condições seja verdadeira, o cpf já é considerado inválido e o valido permanece 0.
- Se o cpf passar por essa primeira verificação, o código segue para o cálculo do primeiro dígito. Caso o resultado seja igual a 10, ele é convertido para 0, já que um dígito verificador só pode ter um único caractere de 0 a 9.
- O mesmo se repete para o segundo dígito verificador, na linha "soma2 = sum(int(cpf_remover[i])*(11 - i) for i in range(10))" -  mas dessa vez considerando os 10 primeiros dígitos do cpf incluindo o primeiro dígito verificador que foi calculado, com pesos decrescentes de 11 até 2 por isso os pesos começam em 11 em vez de 10 nessa segunda conta.
```python
def cpf(cpf_remover):


  valido = 0

  cpf_remover = "".join([ i for i in cpf_remover if i.isdigit()]) # remover os pontos e espacos

  if len(cpf_remover) != 11 or cpf_remover == cpf_remover[0] * 11: # verificar o tamanho do cpf digitado e se numeros se repetem
    valido = 0 # ele fica inválido se na verificação os números se repetirem, que ele cai no 'cpf inválido'.
  else:

    #calcular o primeiro digito verificar

    soma1 = sum(int(cpf_remover[i])*(10 - i) for i in range(9)) # aqui ele soma o numeros do cpf formatado, multiplica por cada número de forma decrescente de 10 até 2
    digito1 = (soma1 * 10) % 11 # aqui ele divide a soma por 11
    if digito1 == 10:# verifica se o 10º número é igual a 10, caso seja ele vira 0.
      digito1 = 0

    soma2 = sum(int(cpf_remover[i])*(11 - i) for i in range(10))# aqui ele soma o numeros do cpf formatado, multiplica por cada número de forma decrescente de 11 até 2
    digito2 = (soma2 * 10) % 11 # aqui ele divide a soma por 11
    if digito2 == 10: # verifica se o 10º número é igual a 10, caso seja ele vira 0.
      digito2 = 0



    valido = digito1 == int(cpf_remover[9]) and digito2 == int(cpf_remover[10]) # ele valida os 2 digitos
  return valido


cpfDigitado = input("Digite seu CPF: ")
valido = cpf(cpfDigitado)

if valido:
  print("cpf válido")
else:
  print("cpf inválido")
```

#Exercicio 37: Implemente uma função recursiva para calcular o fatorial de um número.

---

#Justificativa

Como pede uma função, eu inicio o código com def para declarar uma regra que recebe o fatorial e na função ela espera um objeto do tipo inteiro e que deve retornar algum valor do mesmo tipo.
 - A condição if ela é uma condição de parada sem ela a função iria rodar em um loop e ela retorna 1 para devolver.
 - "return n * fatorial(n - 1)" - nesse trecho estou implementando a recursão que multiplica o número atual pelo resultado da mesma função que ele usa o mesmo número anterior e faz isso até que chegue a 1.
 - E para mostrar o fatorial para o usuário criei a variável "fatorialMostrar" que recebe o nome da função e o número inteiro digitado pelo usuário.




```python


numeroF = int(input("Digite um numero para calcular o fatorial: "))

def fatorial(n: int)-> int: # criando a função de inteiro e devolve um número inteiro


  #caso base
  if n == 1:
    return 1
  # caso recursivo
  return n * fatorial(n -1) # o numero que está recebendo o n, vezes o seu antecessor até que chegue em 1 e retorne o seu valor


fatorialMostrar = fatorial(numeroF) # executando a função com o número digitado
print(fatorialMostrar)


```

 - E depois para a inserção o trecho "lista[n+1] = elementoAtual:" Coloca o número atual no espaço correto que ficou vazio.
 - E por fim utilizo a função format() para pegar o na string que tiver com "{}" e substituir pela variável subsequente.

Inicio um número inteiro, para o usuário digitar, crio uma variável soma que nele utilizo a função sum() para somar todos os elementos nele contido, primeiro eu penso em passar por cada número e o primeiro que se executa é o str(numero) que transforma em string pq não consegui iterar diretamente sobre o número e depois for i in str(numero) passa por cada caractere e o "i" assume os números e então a função sum() soma todos os elementos.


```python
numero = int(input("Digite um numero: "))

soma = sum(int (i) for i in str(numero)) # pegando os numero digitado e somando entre eles
print("A soma dos digitos é: {}".format(soma))


```

#Exercicio 50: Calcule a frequência de cada letra em uma string fornecida pelo usuário.

---

##Justificativa

Comecei iniciando o texto pedindo para o usuário digitar uma palavra, e inicializo a variável frequência com o dicionário vazio.
 - Utilizei o laço for para passar por cada posição do texto utilizando a função len() e depois verificar se a letra já existe, usando o if para verificar se cada texto já existe no dicionário/frequência e depois aqui nesse trecho "frequencia[texto[i]] += " somar mais um se a letra já estiver lá e depois aqui "frequencia[texto[i]] = 1" adiciona a letra com o valor um se ela aparecer pela primeira vez.


```python

texto = input("Digite alguma palavra: ")

frequencia = {}

for i in range(len(texto)):
  if texto[i] in frequencia: # mostra o total de vezes que a letra apareceu e o guarda na frequencia
    frequencia[texto[i]] += 1 # guarda a letra como chave e o total de vezes que ela aparece como valor
  else:
    frequencia[texto[i]] = 1 # e se a letra não aparece mais de uma vez, ela guarda 1

print(frequencia)



```
