Exercicio 21: Implemente a busca binária para encontrar um elemento em uma lista ordenada

```python
numeros = [1, 4, 7, 12, 15, 21, 22, 27, 33, 35, 64, 66, 72, 75]
print("Data a lista {} encontre seu elemento: ".format(numeros))
encontrado = -1 # começa a variável como false
inicio = 0 # o começo da lista
fim = len(numeros) - 1 # define o final da lista
elemento = int(input("Digite o elemento que deseja encontrar: "))

while inicio <= fim: # loop que se repete enquanto o valor do 'inicio' for menor ou igual ao valor do 'fim'
  meio = (inicio + fim) //2 # ele encontra o indice central

  if numeros[meio] == elemento:
    encontrado = meio# se o numero do meio for encontrado ele interrompe após achar
    break
  elif numeros[meio] < elemento:# ele é verdadeiro se o numero do meio for menor
    inicio = meio + 1
  else: # e aqui ele é verdadeiro se o numero do meio é maior
    fim = meio - 1

if encontrado != -1: # ele verifica se a variável mudou de posição
  print("Elemento encontrado! Sua posição na lista é:  {}".format(encontrado))
else:
  print("elemento não encontrado!")
```

Exercicio 22: Crie uma função para criptografar uma mensagem usando a cifra de César.

```python

texto = input("Digite uma frase para criptografar: ")
chave = int(input("Digite a chave de criptografia: "))

texto_criptografado = ""

for letra in texto:
  if letra.isalpha(): # verificar se na letra contém apenas letras do alfabeto
    if letra.isupper(): # verificar se estão em maiusculas
      letra_criptografada = chr((ord(letra) - 65 + chave) % 26 + 65) # verifica a letra na tabela de codigo ASCII do computador e altera o valor da letra, recomeçando com 0, mantém o resultado dentro do limite de 26 letras do alfabeto + o seu valor
    else:
      letra_criptografada = chr((ord(letra) - 97 + chave) % 26 + 97) # tranforma a letra no numero correspondente no alfabeto, e com a chave move a letra para frente ou para tras para fazer a troca.
    texto_criptografado += letra_criptografada # junta a nova letra no final do texto final
  else:
    texto_criptografado += letra
print("Mensagem criptografada: {}".format(texto_criptografado))
```

Exercicio 31: Implemente o algoritmo de ordenação por bolha (Bubble Sort). - comparar pares de elementos vizinhos e os trocar de lugar se tiverem na ordem errada

```python

lista = [3,4,2,5,6,7,8] # lista desordenada

numero = len(lista) # contar quantos itens a lista possui

for i in range(numero):
  for j in range(numero - i - 1): # percorrer os elementos vizinhos da lista o -1 evita olhar um elemento que ja existe
    if lista[j] > lista[j+1]: # comparar o numero atual com o da frente
      lista[j], lista[j+1] = lista[j+1], lista[j] #inverte os valores e o maior vai para a direita deixando a lista ordenada

print(lista)
```

```python


jogador1 = input("Digite o nome do jogador 1 (X): ")
jogador2 = input("Digite o nome do jogador 2 (O): ")

tabuleiro = [" " for i in range(9)] # inicia o tabuleiro vazio com numeros de 1 a 9
turno = 0 # o turno começa em 0

while turno < 9: # enquanto o turno atual for menor que 9 ele continua
  print({tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]})
  print("---------")
  print({tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]})
  print("---------")
  print({tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]})

  if turno % 2 == 0: # definir de quem é a vez, se o turno da vez for para jogador1, caso contrario jogador 2
    jogadorAgora = jogador1
    simbolo = "x"
  else:
    jogadorAgora = jogador2
    simbolo = "o"


  posicaoEscolha = input("{} - {} escolha uma posição de 1 a 9: ".format(jogadorAgora, simbolo)) # jogador escolhe alguma posição

  posicao = int(posicaoEscolha) -1


  if tabuleiro[posicao] == "x" or tabuleiro[posicao] == "o": #verificar se o lugar escolhido esta livre
    print("Essa posição já está ocupada")
    continue



  tabuleiro[posicao] = simbolo #marcar a jogada que deu certo
  turno += 1



  #condicao de vitoria, se o jogador preencher essas 3 posicoes com a mesma marca ele vence
  vitoria = [
      (0,1,2),
      (3,4,5),
      (6,7,8),
      (0,3,6),
      (1,4,7),
      (2,5,8),
      (0,4,8),
      (2,4,6)
  ]

  venceu = False

  for i in vitoria:
    if tabuleiro[i[0]] == tabuleiro[i[1]] == tabuleiro[i[2]] != " ": # passar por cada lista de jogadas e comparar se as tres posicoes guardam a mesma coisa e ver se estao vazias
      venceu = True # marcar que o jogador atual venceu
      break

  if venceu:
    print({tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]})
    print("---------")
    print({tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]})
    print("---------")
    print({tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]})
    print("O jogador {} venceu!".format(jogadorAgora))
    break

  if turno == 9: # se o turno chegar a 9 jogadas entao é empate
    print("Empate")
    break

```

Exercicio 37: Implemente uma função recursiva para calcular o fatorial de um número.

```python


numeroF = int(input("Digite um numero para calcular o fatorial: "))

def fatorial(n: int)-> int: # criando a função de inteiro e devolve um número inteiro


  #caso base
  if n == 1:
    return 1
  # caso recursivo
  return n * fatorial(n -1) # o numero que está recebendo o n, vezes o seu antecessor até que chegue em 1 e retorne o seu valor


fatorialMostrar = fatorial(numeroF) # executando a função com o número digitado
print(fatorialMostrar)


```

Exercicio 46: Calcule a média e o desvio padrão de uma lista de números.

```python

lista = [2, 4, 4, 4, 5, 5, 7, 9]

media = sum(lista) / len(lista) # soma os numeros da lista e divide com sua quantidade

variancia = sum((i - media) ** 2 for i in lista ) / len(lista) # ele ve a distância entre o numero e a média elevado ao quadrado

desvioPadrao = variancia ** 0.5 # calculando o desvio padrão / raiz quadrada

print("A média é: {}".format(media))
print(" O desvio padão é: {}".format(desvioPadrao))


```

Exercicio 48: Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python

lista = [1,2,3,4,53,2,11,4,2,11,3] # lista sem ser ordenada

print("lista desordenada: {}".format(lista))

for i in range(1, len(lista)): # começa o loop a partir do segundo elemento o indice 1
  meio = elementoAtual = lista[i] # aqui ele salva o numero atual e o elementoAtual guarda o número da vez

  n = i -1 # aqui ele aponta para o elemento anterior

  #mover os elementos maiores que o meio para a direita
  while n >= 0 and lista[n] > meio: # loop que continua até não chegar ao início da lista - enquanto o elemento anterior o lista[n] for maior que o meio.
    lista[n+1] = lista[n] # aqui ele move o elemento maior para a adireita
    n -= 1 # tenta testar os elementos anteriores da esquerda
  lista[n+1] = elementoAtual # e aqui ele insere o elementoAtual na posição correta.

print("lista ordenada: {}".format(lista))
```

Exercicio 50: Calcule a frequência de cada letra em uma string fornecida pelo usuário.

```python

texto = input("Digite alguma palavra: ")

frequencia = {}

for i in range(len(texto)):
  if texto[i] in frequencia: # mostra o total de vezes que a letra apareceu e o guarda na frequencia
    frequencia[texto[i]] += 1 # guarda a letra como chave e o total de vezes que ela aparece como valor
  else:
    frequencia[texto[i]] = 1 # e se a letra não aparece mais de uma vez, ela guarda 1

print(frequencia)



```
